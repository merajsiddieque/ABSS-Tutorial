```python
from ddgs import DDGS
from insightface.app import FaceAnalysis
from PIL import Image
import pandas as pd
from pathlib import Path
from tqdm import tqdm
import requests
import cv2
import json
import time

# =====================================================
# Load Metadata
# =====================================================

subjects = pd.read_csv(SUBJECTS_CSV)

REAL_DIR = BASE_DIR / "data/raw/REAL"
REAL_DIR.mkdir(parents=True, exist_ok=True)

# =====================================================
# InsightFace Detector
# =====================================================

app = FaceAnalysis(
    name="buffalo_l",
    providers=["CPUExecutionProvider"]
)
app.prepare(ctx_id=0, det_size=(640,640))

HEADERS = {
    "User-Agent": "Mozilla/5.0"
}

# =====================================================
# Download Image
# =====================================================

def download_image(url, save_path):
    try:
        r = requests.get(url, headers=HEADERS, timeout=15)

        if r.status_code != 200:
            return False

        with open(save_path, "wb") as f:
            f.write(r.content)

        Image.open(save_path).verify()
        return True

    except Exception:
        if save_path.exists():
            save_path.unlink()
        return False


# =====================================================
# Detect + Crop Single Face (256×256)
# =====================================================

def crop_single_face(path):

    img = cv2.imread(str(path))

    if img is None:
        return False

    faces = app.get(img)

    if len(faces) != 1:
        return False

    h, w = img.shape[:2]

    x1, y1, x2, y2 = map(int, faces[0].bbox)

    x1 = max(0, x1)
    y1 = max(0, y1)
    x2 = min(w, x2)
    y2 = min(h, y2)

    if x2 <= x1 or y2 <= y1:
        return False

    crop = img[y1:y2, x1:x2]

    if crop.size == 0:
        return False

    crop = cv2.resize(crop, (256,256))

    cv2.imwrite(str(path), crop)

    return True


# =====================================================
# Resume Previous Progress
# =====================================================

completed = 0

for _, row in subjects.iterrows():

    folder = REAL_DIR / row.subject_id

    if folder.exists() and len(list(folder.glob("*.jpg"))) >= 2:
        completed += 1

print(f"Already completed : {completed}/{len(subjects)}")


# =====================================================
# Download Remaining Subjects
# =====================================================

new_completed = 0

with DDGS() as ddgs:

    remaining = subjects.iloc[completed:]

    for _, row in tqdm(
        remaining.iterrows(),
        total=len(remaining)
    ):

        folder = REAL_DIR / row.subject_id
        folder.mkdir(parents=True, exist_ok=True)

        existing = sorted(folder.glob("*.jpg"))

        if len(existing) >= 2:
            continue

        saved = len(existing)

        query = f'{row["name"]} celebrity close up face portrait high quality'

        try:
            results = list(
                ddgs.images(
                    query,
                    max_results=20
                )
            )

        except Exception:
            continue

        for item in results:

            temp = folder / "temp.jpg"

            if not download_image(item["image"], temp):
                continue

            if not crop_single_face(temp):
                temp.unlink(missing_ok=True)
                continue

            final = folder / f"{row.subject_id}_{saved+1:03d}.jpg"
            temp.rename(final)

            saved += 1

            if saved == 2:
                break

        if saved == 2:
            new_completed += 1

        time.sleep(0.2)


# =====================================================
# Final Statistics
# =====================================================

total_complete = 0

for _, row in subjects.iterrows():

    folder = REAL_DIR / row.subject_id

    if folder.exists() and len(list(folder.glob("*.jpg"))) >= 2:
        total_complete += 1


# =====================================================
# Save metadata.json
# =====================================================

metadata_records = subjects.to_dict(orient="records")

with open(METADATA_JSON, "w", encoding="utf-8") as f:
    json.dump(metadata_records, f, indent=4, ensure_ascii=False)


# =====================================================
# Save statistics.json
# =====================================================

statistics = {
    "total_subjects": int(len(subjects)),
    "completed_subjects": int(total_complete),
    "total_images": int(total_complete * 2),
    "image_size": "256x256",
    "variants": [
        "REAL",
        "GAUSSIAN",
        "SALT_PEPPER"
    ],
    "gender_distribution":
        subjects["gender"].value_counts().to_dict(),
    "ethnicity_distribution":
        subjects["ethnicity"].value_counts().to_dict()
}

with open(STATISTICS_JSON, "w") as f:
    json.dump(statistics, f, indent=4)


# =====================================================
# Summary
# =====================================================

print("\n" + "="*55)
print("DATASET GENERATION COMPLETED")
print("="*55)

print(f"Newly Downloaded     : {new_completed}")
print(f"Completed Subjects   : {total_complete}/{len(subjects)}")
print(f"Total Images         : {total_complete*2}")

print("\nFiles Saved")
print("subjects.csv   :", SUBJECTS_CSV)
print("metadata.json  :", METADATA_JSON)
print("statistics.json:", STATISTICS_JSON)
print("REAL Dataset   :", REAL_DIR)
```

```python
import cv2
import numpy as np
from pathlib import Path
from tqdm import tqdm

REAL_DIR = BASE_DIR / "data/raw/REAL"
GAUSSIAN_DIR = BASE_DIR / "data/raw/GAUSSIAN"
SP_DIR = BASE_DIR / "data/raw/SALT_PEPPER"

# =====================================================
# Noise Functions
# =====================================================

def add_gaussian_noise(image, sigma=20):
    """
    Gaussian sensor noise
    """
    noise = np.random.normal(0, sigma, image.shape).astype(np.float32)

    noisy = image.astype(np.float32) + noise
    noisy = np.clip(noisy, 0, 255)

    return noisy.astype(np.uint8)


def add_salt_pepper(image, amount=0.02):
    """
    Salt & Pepper transmission noise
    """
    noisy = image.copy()

    h, w = image.shape[:2]
    total = int(amount * h * w)

    # Salt (white)
    ys = np.random.randint(0, h, total // 2)
    xs = np.random.randint(0, w, total // 2)
    noisy[ys, xs] = 255

    # Pepper (black)
    ys = np.random.randint(0, h, total // 2)
    xs = np.random.randint(0, w, total // 2)
    noisy[ys, xs] = 0

    return noisy


# =====================================================
# Generate Noise Dataset
# =====================================================

total_images = 0

subjects = sorted([d for d in REAL_DIR.iterdir() if d.is_dir()])

for subject in tqdm(subjects):

    gaussian_subject = GAUSSIAN_DIR / subject.name
    sp_subject = SP_DIR / subject.name

    gaussian_subject.mkdir(parents=True, exist_ok=True)
    sp_subject.mkdir(parents=True, exist_ok=True)

    for img_path in subject.glob("*.jpg"):

        img = cv2.imread(str(img_path))

        if img is None:
            continue

        gaussian = add_gaussian_noise(img, sigma=20)
        saltpepper = add_salt_pepper(img, amount=0.02)

        cv2.imwrite(str(gaussian_subject / img_path.name), gaussian)
        cv2.imwrite(str(sp_subject / img_path.name), saltpepper)

        total_images += 1


print("="*50)
print("NOISE DATASET GENERATED")
print("="*50)
print(f"REAL images processed : {total_images}")
print(f"Gaussian dataset      : {GAUSSIAN_DIR}")
print(f"Salt & Pepper dataset : {SP_DIR}")
```

```python
import cv2
import numpy as np
from pathlib import Path
from tqdm import tqdm

# =====================================================
# Input & Output Directories
# =====================================================

RAW_DIR = BASE_DIR / "data/raw"
PRE_DIR = BASE_DIR / "data/preprocessed"

variants = ["REAL", "GAUSSIAN", "SALT_PEPPER"]

# =====================================================
# CLAHE + Normalize
# =====================================================

clahe = cv2.createCLAHE(
    clipLimit=2.0,
    tileGridSize=(8, 8)
)

def preprocess_image(image):
    # 1. Ensure 256×256
    image = cv2.resize(image, (256, 256))

    # 2. Convert BGR → LAB
    lab = cv2.cvtColor(image, cv2.COLOR_BGR2LAB)

    l, a, b = cv2.split(lab)

    # 3. CLAHE on L channel
    l = clahe.apply(l)

    lab = cv2.merge((l, a, b))

    enhanced = cv2.cvtColor(lab, cv2.COLOR_LAB2BGR)

    # 4. Normalize intensity
    normalized = cv2.normalize(
        enhanced,
        None,
        alpha=0,
        beta=255,
        norm_type=cv2.NORM_MINMAX
    )

    return normalized

# =====================================================
# Process All Variants
# =====================================================

total = 0

for variant in variants:

    input_root = RAW_DIR / variant
    output_root = PRE_DIR / variant

    output_root.mkdir(parents=True, exist_ok=True)

    subjects = [d for d in input_root.iterdir() if d.is_dir()]

    for subject in tqdm(subjects, desc=variant):

        out_subject = output_root / subject.name
        out_subject.mkdir(parents=True, exist_ok=True)

        for img_path in subject.glob("*.jpg"):

            img = cv2.imread(str(img_path))

            if img is None:
                continue

            processed = preprocess_image(img)

            cv2.imwrite(
                str(out_subject / img_path.name),
                processed
            )

            total += 1

print("=" * 50)
print("PREPROCESSING COMPLETED")
print("=" * 50)
print(f"Total images processed : {total}")
print(f"Saved to : {PRE_DIR}")
```

```python
import cv2
import numpy as np
from pathlib import Path
from tqdm import tqdm
from keras_facenet import FaceNet
from skimage.feature import local_binary_pattern
from sklearn.preprocessing import normalize

# =====================================================
# Paths
# =====================================================

PRE_DIR = BASE_DIR / "data/preprocessed"
FEATURE_DIR = BASE_DIR / "data/features"

variants = ["REAL", "GAUSSIAN", "SALT_PEPPER"]

# =====================================================
# FaceNet Model
# =====================================================

embedder = FaceNet()

# =====================================================
# Handcrafted Feature
# =====================================================

def extract_lbp_hist(image):

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    lbp = local_binary_pattern(
        gray,
        P=8,
        R=1,
        method="uniform"
    )

    lbp_hist, _ = np.histogram(
        lbp.ravel(),
        bins=59,
        range=(0,59)
    )

    lbp_hist = lbp_hist.astype(np.float32)
    lbp_hist /= (lbp_hist.sum() + 1e-6)

    hist = []

    for ch in range(3):
        h = cv2.calcHist([image],[ch],None,[32],[0,256])
        h = cv2.normalize(h,h).flatten()
        hist.extend(h)

    hist = np.array(hist,dtype=np.float32)

    feature = np.concatenate([lbp_hist,hist])

    feature = normalize(feature.reshape(1,-1))[0]

    return feature


# =====================================================
# FaceNet Feature
# =====================================================

def extract_facenet(image):

    rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    embedding = embedder.embeddings([rgb])[0]

    embedding = normalize(embedding.reshape(1,-1))[0]

    return embedding


# =====================================================
# Extract Features
# =====================================================

for variant in variants:

    print(f"\nProcessing {variant}")

    lbp_features = {}
    facenet_features = {}

    variant_dir = PRE_DIR / variant

    for subject in tqdm(sorted(variant_dir.iterdir())):

        if not subject.is_dir():
            continue

        for img_path in sorted(subject.glob("*.jpg")):

            img = cv2.imread(str(img_path))

            if img is None:
                continue

            key = f"{subject.name}/{img_path.name}"

            lbp_features[key] = extract_lbp_hist(img)
            facenet_features[key] = extract_facenet(img)

    out_dir = FEATURE_DIR / variant
    out_dir.mkdir(parents=True, exist_ok=True)

    np.save(out_dir / "lbp_hist_features.npy",
            lbp_features,
            allow_pickle=True)

    np.save(out_dir / "facenet_features.npy",
            facenet_features,
            allow_pickle=True)

print("\nFeature extraction completed.")
```

```python
import numpy as np
import pandas as pd
import random
from pathlib import Path
from sklearn.metrics.pairwise import cosine_similarity

# =====================================================
# Paths
# =====================================================

FEATURE_DIR = BASE_DIR / "data/features"
MATCH_DIR = BASE_DIR / "outputs/reports/pair_matching"

MATCH_DIR.mkdir(parents=True, exist_ok=True)

variants = ["REAL", "GAUSSIAN", "SALT_PEPPER"]

subjects = pd.read_csv(SUBJECTS_CSV)

random.seed(42)

# =====================================================
# Generate Genuine & Impostor Pairs
# =====================================================

for variant in variants:

    print(f"\nProcessing {variant}")

    features = np.load(
        FEATURE_DIR / variant / "facenet_features.npy",
        allow_pickle=True
    ).item()

    results = []

    # -----------------------------
    # Genuine Pairs (200)
    # -----------------------------

    for _, row in subjects.iterrows():

        sid = row.subject_id

        img1 = f"{sid}/{sid}_001.jpg"
        img2 = f"{sid}/{sid}_002.jpg"

        if img1 not in features or img2 not in features:
            continue

        score = cosine_similarity(
            [features[img1]],
            [features[img2]]
        )[0][0]

        results.append({
            "pair_type": "genuine",
            "subject_A": sid,
            "subject_B": sid,
            "score": float(score)
        })

    # -----------------------------
    # Impostor Pairs (1000)
    # -----------------------------

    ids = subjects.subject_id.tolist()

    generated = set()

    while len(generated) < 1000:

        a, b = random.sample(ids, 2)

        pair = tuple(sorted([a, b]))

        if pair in generated:
            continue

        generated.add(pair)

        img1 = f"{a}/{a}_001.jpg"
        img2 = f"{b}/{b}_001.jpg"

        if img1 not in features or img2 not in features:
            continue

        score = cosine_similarity(
            [features[img1]],
            [features[img2]]
        )[0][0]

        results.append({
            "pair_type": "impostor",
            "subject_A": a,
            "subject_B": b,
            "score": float(score)
        })

    # Save CSV
    df = pd.DataFrame(results)

    csv_path = MATCH_DIR / f"{variant.lower()}_pairs.csv"

    df.to_csv(csv_path, index=False)

    print(f"Genuine  : {(df.pair_type=='genuine').sum()}")
    print(f"Impostor : {(df.pair_type=='impostor').sum()}")
    print(f"Saved    : {csv_path}")

print("\nPair matching completed.")
```

```python
import numpy as np
import pandas as pd
import json
from pathlib import Path
from sklearn.metrics import roc_auc_score
import matplotlib.pyplot as plt

# =====================================================
# Paths
# =====================================================

PAIR_DIR = BASE_DIR / "outputs/reports/pair_matching"
REPORT_DIR = BASE_DIR / "outputs/reports"
FIGURE_DIR = BASE_DIR / "outputs/figures"

REPORT_DIR.mkdir(parents=True, exist_ok=True)
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

metadata = pd.read_json(METADATA_JSON)

variants = ["REAL","GAUSSIAN","SALT_PEPPER"]

evaluation = {}

# =====================================================
# Evaluation Function
# =====================================================

def evaluate(scores, labels):

    thresholds = np.arange(0,1.001,0.001)

    fmr=[]
    fnmr=[]

    scores=np.array(scores)
    labels=np.array(labels)

    for t in thresholds:

        impostor=(labels==0)
        genuine=(labels==1)

        far=((scores[impostor]>=t).sum())/impostor.sum()
        frr=((scores[genuine]<t).sum())/genuine.sum()

        fmr.append(far)
        fnmr.append(frr)

    fmr=np.array(fmr)
    fnmr=np.array(fnmr)

    idx=np.argmin(np.abs(fmr-fnmr))

    eer=float((fmr[idx]+fnmr[idx])/2)

    auc=float(roc_auc_score(labels,scores))

    # Operational thresholds
    t001=float(thresholds[np.argmin(np.abs(fmr-0.001))])
    t0001=float(thresholds[np.argmin(np.abs(fmr-0.0001))])

    return {
        "thresholds":thresholds,
        "fmr":fmr,
        "fnmr":fnmr,
        "eer":eer,
        "eer_threshold":float(thresholds[idx]),
        "auc":auc,
        "t_0_1":t001,
        "t_0_01":t0001
    }

# =====================================================
# Evaluate Each Variant
# =====================================================

for variant in variants:

    df=pd.read_csv(PAIR_DIR/f"{variant.lower()}_pairs.csv")

    labels=(df.pair_type=="genuine").astype(int)

    result=evaluate(df.score,labels)

    # -----------------------------------------
    # Fairness Analysis
    # -----------------------------------------

    threshold=result["eer_threshold"]

    fairness={}

    overall_fnmr=((df[(labels==1)].score<threshold).sum())/(labels==1).sum()

    for group in ["gender","ethnicity"]:

        stats={}

        for value in metadata[group].unique():

            ids=metadata[metadata[group]==value].subject_id.tolist()

            genuine=df[
                (df.pair_type=="genuine") &
                (df.subject_A.isin(ids))
            ]

            if len(genuine)==0:
                continue

            fnmr=((genuine.score<threshold).sum())/len(genuine)

            ddr=float(fnmr/(overall_fnmr+1e-8))

            stats[value]={
                "FNMR":float(fnmr),
                "DDR":ddr
            }

        fairness[group]=stats

    # Save report
    evaluation[variant]={
        "EER":result["eer"],
        "AUC":result["auc"],
        "EER_THRESHOLD":result["eer_threshold"],
        "FMR_0.1_THRESHOLD":result["t_0_1"],
        "FMR_0.01_THRESHOLD":result["t_0_01"],
        "FAIRNESS":fairness
    }

    # -----------------------------------------
    # ROC / DET Plot
    # -----------------------------------------

    plt.figure(figsize=(6,5))
    plt.plot(result["fmr"],1-result["fnmr"],label=variant)
    plt.xlabel("False Match Rate")
    plt.ylabel("True Match Rate")
    plt.title(f"ROC Curve - {variant}")
    plt.grid(True)
    plt.legend()
    plt.savefig(FIGURE_DIR/f"roc_{variant.lower()}.png",dpi=300)
    plt.close()

    # -----------------------------------------
    # FMR vs FNMR
    # -----------------------------------------

    plt.figure(figsize=(6,5))
    plt.plot(result["thresholds"],result["fmr"],label="FMR")
    plt.plot(result["thresholds"],result["fnmr"],label="FNMR")
    plt.axvline(result["eer_threshold"],ls="--",c="red")
    plt.xlabel("Threshold")
    plt.ylabel("Rate")
    plt.title(f"EER Curve - {variant}")
    plt.legend()
    plt.grid(True)
    plt.savefig(FIGURE_DIR/f"eer_{variant.lower()}.png",dpi=300)
    plt.close()

# =====================================================
# Save JSON Report
# =====================================================

with open(REPORT_DIR/"evaluation_results.json","w") as f:
    json.dump(evaluation,f,indent=4)

print("="*55)
print("EVALUATION COMPLETED")
print("="*55)

for v in variants:
    print(f"{v:12} | EER = {evaluation[v]['EER']:.4f} | AUC = {evaluation[v]['AUC']:.4f}")

print("\nSaved:")
print(REPORT_DIR/"evaluation_results.json")
print(FIGURE_DIR)
```

```python
import numpy as np
import pandas as pd
import json
from pathlib import Path
from sklearn.metrics import roc_auc_score
import matplotlib.pyplot as plt

# =====================================================
# Google Drive Paths (OUTPUT1)
# =====================================================

DRIVE_DIR = Path("/content/drive/MyDrive/Colab Notebooks/ABSS")

PAIR_DIR   = DRIVE_DIR / "output1/reports/pair_matching"
REPORT_DIR = DRIVE_DIR / "output1/reports"
FIGURE_DIR = DRIVE_DIR / "output1/figures"

REPORT_DIR.mkdir(parents=True, exist_ok=True)
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

variants = ["REAL", "GAUSSIAN", "SALT_PEPPER"]
evaluation = {}

# =====================================================
# Evaluation Function
# =====================================================

def evaluate(scores, labels):

    thresholds = np.arange(0, 1.001, 0.001)

    scores = np.asarray(scores)
    labels = np.asarray(labels)

    fmr, fnmr = [], []

    for t in thresholds:

        genuine = labels == 1
        impostor = labels == 0

        far = (scores[impostor] >= t).mean()   # False Match Rate
        frr = (scores[genuine] < t).mean()     # False Non-Match Rate

        fmr.append(far)
        fnmr.append(frr)

    fmr = np.array(fmr)
    fnmr = np.array(fnmr)

    idx = np.argmin(np.abs(fmr - fnmr))

    return {
        "thresholds": thresholds,
        "fmr": fmr,
        "fnmr": fnmr,
        "eer": float((fmr[idx] + fnmr[idx]) / 2),
        "eer_threshold": float(thresholds[idx]),
        "auc": float(roc_auc_score(labels, scores))
    }

# =====================================================
# Evaluate All Variants
# =====================================================

for variant in variants:

    df = pd.read_csv(PAIR_DIR / f"{variant.lower()}_pairs.csv")

    labels = (df["pair_type"] == "genuine").astype(int)

    result = evaluate(df["score"], labels)

    evaluation[variant] = {
        "AUC": round(result["auc"], 4),
        "EER": round(result["eer"], 4),
        "EER_THRESHOLD": round(result["eer_threshold"], 3)
    }

    # ---------------- ROC Curve ----------------

    plt.figure(figsize=(6,5))
    plt.plot(result["fmr"], 1-result["fnmr"], linewidth=2,
             label=f"AUC = {result['auc']:.3f}")
    plt.xlim(0,1)
    plt.ylim(0,1.02)
    plt.xlabel("False Match Rate")
    plt.ylabel("True Match Rate")
    plt.title(f"ROC Curve - {variant}")
    plt.grid(True)
    plt.legend(loc="lower right")
    plt.tight_layout()
    plt.savefig(FIGURE_DIR / f"roc_{variant.lower()}.png", dpi=300)
    plt.close()

    # ---------------- EER Curve ----------------

    plt.figure(figsize=(6,5))
    plt.plot(result["thresholds"], result["fmr"], linewidth=2, label="FMR")
    plt.plot(result["thresholds"], result["fnmr"], linewidth=2, label="FNMR")
    plt.axvline(result["eer_threshold"], color="red", linestyle="--",
                label=f"EER = {result['eer']:.3f}")
    plt.xlim(0,1)
    plt.ylim(0,1.02)
    plt.xlabel("Threshold")
    plt.ylabel("Rate")
    plt.title(f"EER Curve - {variant}")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.savefig(FIGURE_DIR / f"eer_{variant.lower()}.png", dpi=300)
    plt.close()

# =====================================================
# Save Evaluation Report
# =====================================================

with open(REPORT_DIR / "evaluation_results.json", "w") as f:
    json.dump(evaluation, f, indent=4)

print("="*55)
print("OUTPUT1 EVALUATION COMPLETED")
print("="*55)

for k, v in evaluation.items():
    print(f"{k:12} | AUC = {v['AUC']:.4f} | EER = {v['EER']:.4f}")

print("\nReports :", REPORT_DIR)
print("Figures :", FIGURE_DIR)
```
